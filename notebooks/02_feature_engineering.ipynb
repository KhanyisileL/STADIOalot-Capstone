```python
import pandas as pd
import numpy as np
import pickle
```

```python
base_path = '/Users/mohau/Desktop/STADIOalot-Capstone/datasets/'

with open(base_path + 'X_train.pkl', 'rb') as f:
    X_train = pickle.load(f)
with open(base_path + 'X_test.pkl', 'rb') as f:
    X_test = pickle.load(f)
with open(base_path + 'y_train.pkl', 'rb') as f:
    y_train = pickle.load(f)
with open(base_path + 'y_test.pkl', 'rb') as f:
    y_test = pickle.load(f)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
columns = ['hour', 'C1', 'banner_pos', 'site_id', 'site_domain', 'site_category',
           'app_id', 'app_domain', 'app_category', 'device_id', 'device_ip',
           'device_model', 'device_type', 'device_conn_type', 'C14', 'C15',
           'C16', 'C17', 'C18', 'C19', 'C20', 'C21',
           'day_of_week', 'hour_of_day']

X_train_df = pd.DataFrame(X_train, columns=columns)
X_test_df = pd.DataFrame(X_test, columns=columns)

print(X_train_df.head())
print("\nShape:", X_train_df.shape)
```

```python
# Frequency encoding for high-cardinality features
for col in ['device_id', 'device_ip', 'device_model']:
    freq = X_train_df[col].value_counts()
    X_train_df[col + '_count'] = X_train_df[col].map(freq)
    X_test_df[col + '_count']  = X_test_df[col].map(freq).fillna(0)

print("Frequency columns added:")
print("  - device_id_count")
print("  - device_ip_count")
print("  - device_model_count")
print("\nX_train shape now:", X_train_df.shape)
print("X_test shape now: ", X_test_df.shape)
```

```python
with open(base_path + 'X_train_engineered.pkl', 'wb') as f:
    pickle.dump(X_train_df, f)
with open(base_path + 'X_test_engineered.pkl', 'wb') as f:
    pickle.dump(X_test_df, f)

print("Engineered data saved successfully!")
print("Final X_train shape:", X_train_df.shape)
print("Final X_test shape:", X_test_df.shape)
```
